# PHASE 1: Dataset Understanding & Preparation

## Step 1.1 — Load and Inspect the Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load datasets
train_df = pd.read_csv("../data/raw/kdd_train.csv")
test_df = pd.read_csv("../data/raw/kdd_test.csv")

print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)

print("\nTrain columns:", train_df.columns.tolist())

print("\nFirst 5 rows:")
display(train_df.head())
```

## Step 1.2 — Identify Column Types


```python
# Categorical columns (text values)
categorical_cols = [
    "protocol_type",
    "service",
    "flag"
]

# Target column
target_col = "labels"

# Numeric columns (all others)
numeric_cols = [
    col
    for col in train_df.columns
    if col not in categorical_cols + [target_col]
]

print("Categorical features:", categorical_cols)
print("Numeric features count:", len(numeric_cols))
print("Target column:", target_col)
```

## Step 1.3 — Check Class Distribution (Training Set)

```python
print("Training class distribution:")
print(train_df["labels"].value_counts())

print("\nTesting class distribution:")
print(test_df["labels"].value_counts())

# Identify classes in test but NOT in train
train_classes = set(train_df["labels"].unique())
test_classes = set(test_df["labels"].unique())

test_only_classes = test_classes - train_classes

print(
    f"\nClasses in TEST but NOT in TRAIN ({len(test_only_classes)}):"
)
print(sorted(test_only_classes))
```

## Step 1.4 — Check Missing Values and Duplicates

```python
print(
    "Missing values in train:",
    train_df.isnull().sum().sum()
)

print(
    "Missing values in test:",
    test_df.isnull().sum().sum()
)

print(
    "Duplicate rows in train:",
    train_df.duplicated().sum()
)

print(
    "Duplicate rows in test:",
    test_df.duplicated().sum()
)
```

## Step 1.5 — Check Feature Ranges

```python
print("Feature ranges (train):")

display(
    train_df[numeric_cols]
    .describe()
    .T[["min", "max", "mean"]]
)
```

## Step 1.6 — Save Prepared Data

```python
# Keep copies for processing
train_raw = train_df.copy()
test_raw = test_df.copy()

print("Phase 1 complete: Dataset loaded and inspected.")
```
